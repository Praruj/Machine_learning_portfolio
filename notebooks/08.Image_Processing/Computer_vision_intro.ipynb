## Image Acquistion Fundamentals in Digital processing

It is process of turning something from real world into numbers that a computer can understand. Light or another type of energy hits an object, some of that energy is reflected or passes through the object, and sensor captures it and converts it into electric signal, which is then turned into digital numbers.



The core of digital image formation is the function $f(x,y)$, which is determined by the illumination source $i(x,y)$, and the reflectance $r(x,y)$ from the scene.

Two important steps:

1. Sampling decides where we take measurements, which decides the number of pixels and therefore spatial resolution (how much detail we can see).  

2. Quantization decides what numerical intensity value we assign to each measurement which determines the intesity resolution (how many brightness levels can we represent).

Dynamic range describes how well the imaging system can capture both very dark and very bright areas. For color images,, we usually have three channels - Red, Green and Blue (RGB) - with each pixel containing a value for each channel.

Once we have image, we use restoration to try to recover an image damged by noise or blur or enhancement to make it look or work better.

Compression reduces the amount od data needed to store the image by taking advantage of repaeated or unnecessary information. For eg; nearby pixels are often similar (spatial redundancy) and consecutive video frames often contain mostly the same information.

Image compressions systems use encoders and decoders which eliminate redundaces through mapping (to reduce/ spatial temporal redundancy), Quantization (to discard irrelevant information) and symbol coding (assigining codes to quantizer output), while decoders reverse the process except quantization.


- Spatial redundancy → nearby pixels are similar.
- Temporal redundancy → nearby video frames are similar.
- Coding redundancy → some values appear much more often than others, so we can give common values shorter codes.

“Why store the same information over and over when I can represent it more efficiently?”

Before building a computer vision model, first understand the **data**.

Ask:

- **What was measured?**
- **How was it measured?**
- **What does each pixel mean?**
- **What coordinate system is being used?**
- **What are the noise characteristics?**
- **What will the model actually see after deployment?**

Then decide whether you need:

- Preprocessing
- Resizing
- Additional channels
- Fine-tuning
- A modified final layer
- A different model

> **Key takeaway:** Good computer vision starts with understanding the data — not immediately choosing a neural network.

## Feature Description

It is a useful property of an image that helps a model recognize something. A descriptor is numerical representation of those useful features.

- Feature Representation

1. Numerical: Can be stored as arrays/Lists. Tensors = Multidiemensional arrays commonly used in ML

2. Categorical features: For categories like dogs,cats etc we can use numerical labels, dictionaries/ lists, one-hot encoding = represent each category as a binary vector.

For example: cat=[0,1,0]

3. Image features: pixel value metrices . features learned extracted by CNNs.

### What makes a Good descriptor?
1. Transformation invariant: should remain useful if the object rotates, moves, changes size, experiences lightening changes.

2. Distinctiveness: different objects should produce sufficiently different descriptors.

3. Reasonable dimensionality: contain emough information without becoming unncessarily huge.

4. Locality: a small region rather than entire image.

### SIFT = Scale invariant Feature Transform

It Finds distinctive local keypoints in an image and creates descriptors for them.
- Find intresting points, describe them and find same point in another image.

1. Step 1: Scale space detection
It searches at different image scsales for interesting points. It uses the difference of Guassians (DoG) to find location where the image changes significantly. These become potential keypoints.

2. Step 2: Keypoint Localization
SIFT improves the keypoints location and removes the unreliable points such as:
- very low contrast points
- points mainly located on edges

3. Step 3 : Orientation
It calcuals the dominant direction of the image gradients around the keypoint. This allows the feature to remain useful when the images is rotated.

4. Step 4 : Descriptor generation
It examines the local region around the keypoint and summarizes its gradient information into a snumerical descriptor.

5. Step 5: Matching
Descriptors from two images are compared.

It is used in
- Object recognition
- image stitching
- 3D reconstruction

## SURF = Speeded up Robust Features

Surf has a similar goal to SIFT but focuses more strongly on computational speed.

1. An Integral image is precomputed representation that makes calculating rectangular regions very fasr. This helps SURF process features efficiently.

2. Blob Detection: SURF uses Hessian matrix to find blob like regions where intensity changes strongly in multiple directions

3. Scale Detection: It examines different scales and use the determinat of Hessian matrix to find stable keypoints

4. Orientation: determines the dominant direction around each keypoint using Haar wavelet responses.

5. Descriptor and matching: The responses are converted into a feature vector, which can be compared with descriptors from another image.

```TEXT
IMAGE
  ↓
Find interesting points
  ↓
Describe those points numerically
  ↓
Compare descriptions
  ↓
Find matching features


SIFT and SURF are classical computer-vision techniques for doing exactly this.

Modern CNNs can learn feature representations automatically,
but understanding SIFT/SURF teaches the fundamental idea
of "turn visual patterns into useful numerical representations."

```
