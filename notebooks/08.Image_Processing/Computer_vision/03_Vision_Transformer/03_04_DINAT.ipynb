# DINAT: Dilated Neighbourhood attention Transformer (DiNAT)

Standard vision transformer tries to compare every single pixel to other pixel. It gives perfect understanding but takes an exhausting amount of time and brainpower(compute).

The DINAT solves this using two steps:

1. Neighbourhood Attention (NA)- "looking close"

Instead of looking whole picture, we only look at the pixel and the immediate pixel touching it. We figure out local details, like realizing we are looking at patch of something, which is highly efficient, but we cant tell if the patch belongs to some classes or not since we lack big picture.



2. Dilated Neighbourhood Attention (DiNA) - "looking wide"
to see the big picture withour doing extra work, We "dilate" or "expand" our view. We still only look at the same number of pixel but we space them out. We look at one pxiel, skip two and look at next, skip two. Suddenly, our small tube captures the outline of bear ear and nose without needing to process every single pixel in between.


DiNAt alternates between these two methods. It looks closely to get the details (NA), then spaces its view out to get the overall shape (DiNA).

* The core problem is Global self attention in standard vision Transformer scales quadriatically with image size, making them too heavy for high resolution images.

* The Dinat solution introduces a hierarchical structure that combines local attention (NA) with sparse, spread out attention (DiNA).

* Performance:Expanding the receptive field (how much of the image the mode considers at once) without adding extra math operations allows DiNAT to outperform models like Swin and ConvNext.

* Implementation: Acts as backbone for CV tasks (classification, object detection, segmentation)



```python
from transformers import AutoImageProcessor, DinatForImageClassification
from PIL import Image
import requests

url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)

feature_extractor = AutoImageProcessor.from_pretrained("shi-labs/dinat-mini-in1k-224")
model = DinatForImageClassification.from_pretrained("shi-labs/dinat-mini-in1k-224")

inputs = feature_extractor(images=image, return_tensors="pt")
outputs=model(**inputs)
logits=outputs.logits

predicted_class_idx = logits.argmax(-1).item()
print("Predicted class:", model.config.id2label[predicted_class_idx])
```
