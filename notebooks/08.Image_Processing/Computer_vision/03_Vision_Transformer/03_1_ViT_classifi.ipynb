# Vision Trasformers : Transfer Learning & Fine-tuning Vision Transformers

## *Vision Transformers*:

Transformers were originally extremely successful in NLP. For example,  

" The cat is sleeping" can be broken into tokens: `the | cat | is | sleeping`.

The transformer processes relationships between these tokens.

" can we treat parts of an image like tokens ?"
> yes, Instead of words,  we use image patches.


2. How does ViT turn an image into tokens ?

imagine:

```
┌─────────────────────────┐
│                         │
│        🐶               │
│                         │
│                         │
└─────────────────────────┘

```

Break it into small patches:

```
┌────┬────┬────┬────┐
│ P1 │ P2 │ P3 │ P4 │
├────┼────┼────┼────┤
│ P5 │ P6 │ P7 │ P8 │
├────┼────┼────┼────┤
│ P9 │P10 │P11 │P12 │
├────┼────┼────┼────┤
│P13 │P14 │P15 │P16 │
└────┴────┴────┴────┘

```
Each patch becomes something similar to a token.

so,

```

IMAGE
  ↓
PATCHES
  ↓
PATCH EMBEDDINGS
  ↓
TRANSFORMER
  ↓
CLASSIFICATION

```
The core idea of Vision Transformer.

> Vision Transformer treats image patches like tokens in transformer.

3. Why patches ?

Transformers dont understand pixel naturally, they understand sequence of vectors/tokens. so we convert

```

Image
 ↓
2D pixels
 ↓
small patches
 ↓
vectors
 ↓
tokens

```

For example an image is 🇰

```
224 × 224

and patch size is:

16 × 16

Then the image contains:
```

$224/16 * 224/16$

= 196 patches

```
so the transformers receives roughly:
196 image tokens, rather than 50,176 individual pixels.

```


4. Difference between CNN vs ViT :

CNNs have built in assumptions about images which is called
>Inductive biases

Inductive biases: An assumption built into a model that helps it learn from a limited data.

 * CNN inductive bias
    * Locality: It assumes that nearby pixels are strongly related. It focus on local neihbourhood which is called locality.

    * translational equivariance: CNNs are naturally good at recognizing the same local feature even when its position changes.

    > A useful feature can be useful feature even its position is moved.

---
  | CNN                                     | Vision Transformer                      |
| --------------------------------------- | --------------------------------------- |
| Strong image-specific inductive biases  | Weaker image-specific inductive biases  |
| Locality built in                       | Locality must largely be learned        |
| Translation-related structure built in  | Must learn more from data               |
| Often performs well with less data      | Benefits greatly from large datasets    |
| Highly effective for traditional vision | Highly scalable with large data/compute |


----


> CNNs have stronger built-in assumptions, while ViTs are more flexible and scalable.


5. Why ViTs eventually outperform CNNs ?

CNNs have useful assumptions:

```

locality
+
translation structure

```
Extremely useful when data is limited, but assumptions can also become limitations. ViTs make fewer image specific assumptions. Therefore, with enough:

```
DATA
+
COMPUTE
+
MODEL CAPACITY

```
ViT can learn useful visual relationships from the data itself.

so the tradeoff is ⁉

```
Small/medium dataset
        ↓
CNN inductive bias helps

Huge dataset
        ↓
ViT can learn more flexible representations
```

6. ViT needs alot of data to learn. so, we use ViT that is already trained and reuse what it learned, which is transfer learning.

7. **Transfer Learning**

Instead of teaching the vision model from scratch, we reuse its existing knowledge:

```
Huge dataset
     ↓
Pretrained ViT
     ↓
learned visual features
     ↓
OUR dataset
     ↓
new classifier
```

> Transfer learning means taking knowledge learned from one task/domain and resuing it for another related task.

8. **Pre Trained Model**

A model whose weights have already been learned on large dataset.

```
Large dataset
      ↓
Training
      ↓
Pretrained ViT
      ↓
Your dataset

```

> reuse learned weights instead of starting from random weights


9. Transfer learning vs Fine tuning

Transfer learning: reuse knowledge from a pretrained model.

Fine tuning: Continue training the pretrained model on dataset.

```
Transfer learning
       │
       ├── Freeze most weights
       │     ↓
       │   train selected layers
       │
       └── Fine-tune many/all weights
             ↓
          adapt model

```

10. Simple transfer learning strategy:

```

┌─────────────────────────┐
│ Pretrained ViT          │
│                         │
│ Transformer layers      │ ← FROZEN
│ Transformer layers      │ ← FROZEN
│ Transformer layers      │ ← FROZEN
│                         │
│ Classification head     │ ← TRAIN
└─────────────────────────┘

```

Only classification head learns because the pretrained model already learned useful visual representations.


Freezing : Dont update its weights during training.


```
Normally:

prediction
 ↓
loss
 ↓
backpropagation
 ↓
weights updated



with a frozen layer:

prediction
 ↓
loss
 ↓
backpropagation
 ↓
❌ weights not updated

```
Layer still performs computation, simply doesnot learn new weights.

Why freeze layers ?

* less computation
* Less GPU memory
* Less risk of overfitting


11. The classification head

Pretrained model may have originally for 1000 classes but our problem might have 10 classes. So we can replace/adapt the final classification layer.

```
Pretrained ViT
      ↓
Feature representation
      ↓
NEW classification head
      ↓
10 classes
```

The backbone provides features. The new head maps those features to our classes.

12. Fine - Tuning:

Suppose the pretrained model learn from ordinary photograph but dataset contains medical images, pretrained features may not transfer perfectly. then simply training final layer is not enough so we can fine-tune more of network.

```

Freeze some layers
        ↓
Fine-tune later layers
        ↓
Train classification head


or

Entire pretrained model
        ↓
Fine-tune all weights

```

Use a smaller learning rate: we dont want training to completely destroy the knowledge. A huge learning rate might cause:

```
pretrained knowledge
       ↓
large updates
       ↓
knowledge gets damaged



instead

pretrained knowledge
       ↓
small updates
       ↓
gradually adapt
       ↓
new domain


Thats the intution behind using lower learning rate during fine tuning.



useful model:

Small dataset + similar domain
        ↓
Freeze most layers

Moderate dataset + somewhat different domain
        ↓
Fine-tune some layers

Large dataset + very different domain
        ↓
Fine-tune more/all layers


```


13. Multi class classification:
Each image belongs to exactly one class.

```
Example:

Image → 🐶

Possible classes:

Dog
Cat
Horse
Bird

But one image gets one label.

Image 1 → Dog
Image 2 → Cat
Image 3 → Horse

the model generally chooses one class
```

14. Multi label classificaton:
One image can have multiple labels simultaneously:

```
For example:

Image
 ├── dog
 ├── grass
 ├── outdoor
 └── person

The labels aren't mutually exclusive.

So:

Multi-class:
Dog OR Cat OR Horse

Multi-label:
Dog AND Grass AND Outdoor

That's a critical distinction.

```

| ------------------------ | ------------------- | ---------------------------------- |
| Basic unit               | Pixels/features     | Image patches/tokens               |
| Locality                 | Built in            | Mostly learned                     |
| Translation-related bias | Built in            | Weaker                             |
| Data efficiency          | Often strong        | Benefits from large datasets       |
| Scalability              | Strong              | Excellent                          |
| Global relationships     | Built progressively | Attention directly connects tokens |
| Pretraining              | Useful              | Extremely important in practice    |


```

1. What is a Vision Transformer?

A Transformer that processes an image as a sequence of patches/tokens.

2. Why divide an image into patches?

Transformers operate naturally on token/vector sequences, so image patches can be converted into token embeddings.

3. What is inductive bias?

An assumption built into a learning algorithm that helps it generalize.

4. What inductive biases are important in CNNs?

Locality and translation-related structure/equivariance.

5. Why can CNNs perform better with limited data?

Their built-in image assumptions reduce how much they need to learn from data.

6. Why do ViTs benefit from massive datasets?

They have fewer image-specific inductive biases, so large datasets allow them to learn visual structure from the data.

7. What is transfer learning?

Using knowledge/weights learned from a pretrained model for a new task.

8. What is fine-tuning?

Continuing to train a pretrained model on a new dataset, usually with a smaller learning rate.

9. Why freeze layers?

To reduce computation/memory, speed up training, preserve pretrained features, and reduce overfitting risk.

10. What is the difference between multi-class and multi-label?
Multi-class → one class per image
Multi-label → multiple classes per image

```


>CNNs bake useful image assumptions into the architecture; ViTs trade some of those assumptions for scalability, which makes pretrained models extremely valuable; transfer learning reuses that pretrained knowledge, while fine-tuning adapts it to your specific dataset.
