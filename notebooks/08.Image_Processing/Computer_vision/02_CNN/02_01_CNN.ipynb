# Convolutional Neural Networks

> **Convolution = Feature Extraction**

* Works with **1D, 2D, or 3D data**
* Uses a small matrix called a **kernel (filter)**
* The kernel **slides across the data**
* At each position:
  **Multiply corresponding values → Sum the products**
* The resulting values form a **feature map**

```text
Data + Kernel → Slide → Multiply & Sum → Feature Map
```

---

## 1D Convolution

**Kernel:** `[-1, 1]`

* Detects the **rate of change** (derivative) in the data.
* Large changes can indicate an **edge**.

### Steps

1. Start at the left-most element.
2. Place the kernel over the data.
3. Multiply overlapping values.
4. Add the results.
5. Move the kernel and repeat.

---

## Padding

The kernel has a **center**. Here, we choose `1` (the right element).

To let the center touch the first data element, we add an imaginary `0` to the left.

This `0` is called **padding**.

```text
Data:      [a  b  c  d]
Padding: [0  a  b  c  d]

Kernel:   [-1  1]
               ↑
             center
```

### Why use padding?

* Allows the kernel to process **boundary/edge elements**
* Helps control the **output/feature-map size**

### Types of padding

* **Valid padding** → no padding → output becomes smaller
* **Same padding** → padding preserves spatial dimensions (with stride 1)

> **Padding = Extra values (often zeros) added around the data to handle edges.**

---

## Stride

**Stride = how far the kernel moves at each step.**

* **Stride 1** → moves 1 position
* **Stride 2** → moves 2 positions
* Larger stride → generally produces a **smaller feature map**

```text
Kernel → Move → Calculate → Move → Calculate → ...
```

> **Stride controls the movement of the kernel.**

---

## Feature Map

The **convolved data** (output of the convolution) is called a **feature map**.

> **Feature Map = representation showing where/how strongly a particular feature was detected.**

Examples:

* Edge filter → feature map showing edges
* Texture filter → feature map showing textures
* Learned CNN filter → feature map showing a learned pattern

---

# 2D Convolution

For images, convolution uses a **2D kernel**, such as `3 × 3`.

### Prewitt Filter

To detect changes/edges in one direction, we can use a **Prewitt filter**:

```text
[-1  0  1]
[-1  0  1]
[-1  0  1]
```

* The kernel center touches each pixel.
* Padding can keep the feature map the **same size** as the original image.
* The output is a **feature map showing detected edges/changes**.

### Direction

The Prewitt operator has separate filters for detecting changes in different directions:

```text
Horizontal changes:

[-1  0  1]
[-1  0  1]
[-1  0  1]


Vertical changes:

[-1 -1 -1]
[ 0  0  0]
[ 1  1  1]
```

> **Sobel filters** are another famous family of edge-detection filters.

---

# Output Size

The convolution output size can be calculated using:

$$
Output = \left\lfloor\frac{N + 2P - K}{S}\right\rfloor + 1
$$

Where:

* `N` = input size
* `P` = padding
* `K` = kernel size
* `S` = stride

For 2D images, calculate the output size separately for **height and width**.

Manually designing filters like **Prewitt** or **Sobel** does not work well for every image.

```text
Input Image
     ↓
Randomly initialized filters
     ↓
Convolution
     ↓
Feature Maps
     ↓
Classifier
     ↓
Prediction
     ↓
Loss
     ↓
Backpropagation
     ↓
Update filters
     ↓
Better filters
```

### Key idea

> **CNNs learn which filters are useful for extracting important features from images.**

The filter values are **parameters/weights** that are optimized during training using **backpropagation + gradient descent**.

# Pooling

Pooling is used to **reduce the size of feature maps while keeping important information**.

Imagine an edge appears across several nearby pixels. We don't necessarily need the exact pixel location — we mainly need to know:

> **"Is this feature present in this region?"**

Pooling helps us keep that important information while reducing the amount of data.

---

## Max Pooling

**Max pooling** takes the largest value from a small region.

For example, with a `2 × 2` pooling window:

```text
[1  3]
[2  5]
```

Take the maximum:

```text
Max = 5
```

Another example:

```text
Feature Map:

[1  3  2  4]
[5  6  1  2]
[7  2  8  3]
[4  1  2  9]
```

Using `2 × 2` max pooling:

```text
[1  3]   [2  4]
[5  6]   [1  2]

   ↓         ↓

   6         4


[7  2]   [8  3]
[4  1]   [2  9]

   ↓         ↓

   7         9
```

Result:

```text
[6  4]
[7  9]
```

So the feature map becomes **smaller**.

---

## Why use Pooling?

### 1. Reduces computation

Smaller feature maps → fewer values to process.

### 2. Keeps important features

Max pooling keeps the **strongest activation** in each region.

### 3. Makes the model more robust

The exact location of a feature becomes less important.

For example, to recognize a face, we don't need the eyes to be at the **exact same pixels** in every image.

We mainly need to detect:

```text
Eyes + Nose + Mouth → Face
```

Pooling helps make the network less sensitive to small shifts and changes.

### 4. Reduces parameters/data

Smaller feature maps mean less information needs to be passed to later layers.

---

# Types of Pooling

Common types include:

* **Max Pooling** → takes the maximum value
* **Average Pooling** → takes the average
* **Weighted Pooling**
* **L2 Pooling**

The most commonly encountered one is **Max Pooling**.

# Important CNN Layers

### `Conv2D`

Performs 2D convolution.

Important parameters include:

* **Number of filters**
* **Kernel size**
* **Activation function**
* **Padding**
* **Stride**

Example conceptually:

```text
Conv2D(
    filters = 32,
    kernel_size = 3×3
)
```

This means the layer learns **32 different filters**, each looking for potentially different patterns.

---

### `MaxPooling2D`

Reduces the spatial size of feature maps.

Important parameter:

```text
pool_size = 2×2
```

It takes the maximum value from each `2 × 2` region.

---

### `Dense`

A fully connected layer used near the end of the network.

Important parameter:

```text
units = number of output neurons
```

---

# Padding in CNNs

There are two common approaches:

### `valid`

No padding is added.

```text
Input → Convolution → Smaller output
```

The kernel cannot fully cover the border pixels, so the spatial dimensions **shrink**.

### `same`

Padding is added so that the output can maintain the spatial dimensions (typically with stride `1`).

```text
Input → Padding → Convolution → Same spatial size
```

---

## ⚠️ Important Point From This Lesson

The example uses **no padding**, so the kernel doesn't touch the outermost pixels in the same way.

Therefore:

```text
Input:       26 × 26
      ↓ Conv2D
Output:      smaller than 26 × 26
```

You lose some border information because the kernel must fit completely inside the image.

````python
from tensorflow import keras
from tensorflow.keras import layers

input_shape = (28, 28, 1)
num_classes = 10
model=keras.Sequential(
    [
        keras.Input(shape=input_shape),
        layers.Conv2D(32,kernel_size=(4,4),activation="relu"),
        layers.MaxPooling2D(pool_size=(2,2)),
        layers.Conv2D(64,kernel_size=(3,3),activation="relu"),
        layers.MaxPooling2D(pool_size=(2,2)),
        layers.Flatten(),
        layers.Dropout(0.5),
        layers.Dense(num_classes,activation="softmax")
    ]
)
model.summary()
````

# CNN — New Concepts Only

## 🧩 CNN Layer Flow

```text
Input
 ↓

Conv2D
 ↓
MaxPooling
 ↓

 ↓

```

### Key new concepts

* **Multiple Conv2D layers:** Each layer receives the **previous layer's feature maps**, not the original image.
* **Increasing filters:** Common pattern:

  ```text
  32 → 64 → 128 → ...
  ```

  Deeper layers learn increasingly complex features.

---

## 📐 Understanding Shapes

For:

```python
Conv2D(32, kernel_size=(4,4))
```

* `32` = number of filters
* `(4,4)` = size of each filter
* Output has **32 feature maps**

Example:

```text
28×28×1
   ↓ Conv2D(32, 4×4)
25×25×32
```

`25 × 25` comes from:

```text
28 - 4 + 1 = 25
```

---

## 🧱 Padding

Padding is **optional**.

```text
valid → no padding
same  → adds padding
```

With `valid`, spatial size shrinks.

With `same` + stride `1`, spatial size stays the same.

---

## 🏃 Stride

**Stride = movement size of the kernel.**

```text
stride 1 → move 1 pixel
stride 2 → move 2 pixels
```

Larger stride → smaller output.

---

## 🔄 Parameter Sharing

The **same kernel weights are reused** as the kernel moves across the image.

> **Same filter, different locations.**

This greatly reduces the number of parameters.

---

## 🔍 Sparse Interactions

A neuron only sees a **small local region** of the image, e.g. `3×3`, rather than the entire image.

> **Small connection area = sparse interaction**

### Don't confuse:

```text
Parameter Sharing → SAME weights
Sparse Interaction → SMALL region
```

---

## 🎯 Training Concept

CNN filters start with random values.

Training repeatedly does:

```text
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradient
   ↓
Update kernel weights
```

> **Goal: find kernel values that extract features useful for the task.**

---

## 🧠 Two Useful Terms

**Downsampling**
→ Reducing spatial dimensions, usually through pooling or stride.

**Receptive Field**
→ The portion of the original input that a neuron can "see."
Deeper CNN neurons generally have a **larger receptive field**.
